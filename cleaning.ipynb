```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the CSV file
df = pd.read_csv('shipment.csv')

# Clean the data by removing single tick symbols from the beginning of the cells
df = df.map(lambda x: x.lstrip("'") if isinstance(x, str) else x)

# Save the cleaned data to a new CSV file


manila_rows = df[(df['portname'] == 'Manila') & (df['year'] == 2024)]

# Plot export_cargo for Manila over time

print(manila_rows.describe())

plt.figure(figsize=(10, 5))
plt.plot(manila_rows['date'], manila_rows['import_cargo'])
plt.title('Export Cargo for Manila Over Time')
plt.xlabel('Date')
plt.ylabel('Export Cargo')
plt.xticks(rotation=45)
plt.tight_layout()
plt.xticks([])
plt.show()



```
