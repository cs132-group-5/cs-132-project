```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Read the CSV file
df = pd.read_csv('shipment.csv')

# 1. Count rows with at least one missing value
rows_with_nan = df.isna().any(axis=1).sum()
print(f"Total rows with missing values: {rows_with_nan}")

# 2. Remove rows with missing values
df_cleaned = df.dropna().copy()

# 3. Clean the data (removing single ticks)
df_cleaned = df_cleaned.map(lambda x: x.lstrip("'") if isinstance(x, str) else x)

# 4. Save the cleaned data
df_cleaned.to_csv('shipment_cleaned.csv', index=False)
print("\nCleaned file saved as 'shipment_cleaned.csv'")
```

```python
manila_rows = df[(df['portname'] == 'Manila') & (df['year'] == 2024)]

# Plot export_cargo for Manila over time

print(manila_rows.describe())

plt.figure(figsize=(10, 5))
plt.plot(manila_rows['date'], manila_rows['import_cargo'])
plt.title('Export Cargo for Manila Over Time')
plt.xlabel('Date')
plt.ylabel('Export Cargo')
plt.xticks(rotation=45)
plt.tight_layout()
plt.xticks([])
plt.show()


```
